
### `sorted()` function in python

The `sorted()` function returns a sorted list of specified iterable object.

`sorted(iterable, key, reverse)`

- `iterable` is a required argument. This is the iterable object to sort (list, dict, tuple, etc)
- `key` is an optional argument. Default is None. The key is basically a function to execute to decide the order.
- `reverse` is an optional boolean argument. Default is False. False => ascending order, True => decending order


````python
a=['apple', 'guave', 'orange','banana', 'pineapple', 'grapes']
````

````python
sorted(a)
sorted_a = sorted(a)
print(sorted_a)
````

````python
a
````

````python
a.sort() # method to sort the list in place
a
````

````python
my_list = [[1,2,3],[4,5,6],[3,2,-1],[-4,5,6]]
sorted(my_list, key=lambda a: sum(a), reverse=True)
````

### User defined function


````python
def add(a,b):
    c = a+b
    return c
````

````python
add(1,2)
````

````python
def my_sort_func(v):
    return len(v)
a=['samiksha','ram','nabin','nirmalss','shubhan']
sorted(a, key = my_sort_func)
````

````python
fruits = {'apple': 120, 'banana': 40, 'pineapple': 80, 'mango': 150}
````

````python
sorted(fruits) #retun a list with sorted keys of the dictionary
````

````python
sorted(fruits.items()) # return key-value pairs sorted on keys
````

````python
sorted(fruits.items(), key=lambda a: a[1]) # return key-value pairs sorted on values
````

````python
sorted_dic = dict(sorted(fruits.items(), key=lambda a: a[1]))
sorted_dic
````

````python
for k in sorted_dic.items():
    print(k)
````

### List comprehension


````python
my_list = [3,5,10,11,2]
# create another list with square values of the my_list\
result_list = []
for x in my_list:
    result_list.append(x**2)
result_list
````

````python
result_list = [x**2 for x in my_list]
result_list
````

````python
number_list = [-3,8,9,-4,0,12,4,5]
# create another list from the element of the number list
# condition: in the new lists, square the number if it less than 0, else leave as it is
new_number_list = [x**2 if x < 0 else x for x in number_list ]
new_number_list
````

````python
values = [3,-2,5,12,9,25,32,-4]
# create a new list from the values where the (element+4)**2-10>0
new_list = [x for x in values if (x+4)**2-10>0]
new_list
````
