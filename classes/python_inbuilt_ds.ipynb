## Set

Set is data structure which contains unique values
Sets are also immutable.
Indexing and slicing are not allowed on set


```python
my_list = [1,8,3,4,1, 4,5]
my_set = set(my_list)
my_set
```

```python
s = set()
s.add(1)
s.add(5)
s.add(3)
s
```

```python
for x in s:
    print(x)
```

```python
A = {3,6,8}
B={2,3,8}
A.difference(B)
```

```python
A.intersection(B)
```

```python
A.symmetric_difference(B)
```
