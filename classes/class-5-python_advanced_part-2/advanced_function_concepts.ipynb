## Advanced functions


### Functions in functions


```python
def count(t):
    def int_count(t):
        return t+1
    def str_count(t):
        return len(t)
    
    if type(t)==int:
        return int_count
    if type(t)==str:
        return str_count
```

```python
f = count(1)
f(2)
```

```python
g = count('hello')
g('hello')
```

## Function as an argument to a function


```python
def say_welcome():
    return 'Hello, welcome to the class.'
```

```python
say_welcome()
```

```python
def make_uppercase(function):
    def upper():
        return_string = function()
        uppercase_string = return_string.upper()
        return uppercase_string
    return upper
```

```python
f = make_uppercase(say_welcome)
f
```

```python
f()
```

```python
def increment(n):
    def add():
        return n+1
    n = n+2
    return add()
```

```python
increment(2)
```

## Decorators


```python
def say_bye():
    return "bye, bye, see you soon"
say_bye()
```

```python
@make_uppercase
def say_be_updated():
    return 'bye, bye, see you soon updated'
say_be_updated()
```

### Decorators with arguments


```python
def square(function):
    def wrapper(a,b,c):
        value = function(a,b,c)
        return value **2
    return wrapper
```

```python
@square
def add(n1,n2,n3):
    return n1+n2+n3
```

```python
add(1,2,3)
```

```python
with open('./sample2.txt', 'r') as file:
    content = file.read()
    file.seek()
    print(content)
```

```python
try:
    number_file = open('numbers.txt', 'r');
    sum = 0
    number_count = 0
    for x in number_file:
        sum += float(x)
        number_count += 1
    print(f'Sum of numbers: {sum} and average : {sum/number_count}')
except Exception as e:
    print(e)
finally:
    number_file.close()
```

```python
with open('./sample.txt', 'r+') as file:
    
    l = 0
    for i in file:
        l+=1
        if l==2:
            u = i.replace('senior', 'junior')
            file.write(u)
file.close()
    
```
