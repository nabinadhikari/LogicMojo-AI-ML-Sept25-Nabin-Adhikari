### Native data types

There are primarily 4 types of data in python

- integer (int)
- floating point (float)
- boolean (bool)
- string (str)

Naming convention

- Cannot contain space
- Cannot contain any special character except \_
- It can start or end with underscore
- It can contain alphanumeric character
- It cannot start with numbers
- It cannot be reserved keywords


```python
e = 2
print(type(e))
e = 4.3
print(type(e))
```
